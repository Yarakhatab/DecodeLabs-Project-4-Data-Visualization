```python
import pandas as pd
import numpy as np
```

```python
df=pd.read_csv('/content/Dataset for Data Analytics(Sheet1).csv')
```

```python
info_direct= {
    "Total Rows": df.shape[0]
    ,"Total Columns": df.shape[1]
    ,"Missing Values": df.isnull().sum()
    ,"Duplicate Rows": df.duplicated().sum()
}
print("--- Data Info ---")
for k, v in info_direct.items():
    print(f"{k}: {v}")
```

```python
print('Data Types')
df.dtypes
```

```python
df['Date']=pd.to_datetime(df['Date'])
df['YearMonth']=df['Date'].dt.to_period('M').astype(str)
df['Year']=df['Date'].dt.year
df['Month Name']=df['Date'].dt.month_name()
df['WeekOfMonth']=df['Date'].dt.isocalendar().week # Updated to use .isocalendar().week as .weekofyear is deprecated.
df['DayName']=df['Date'].dt.day_name()
```

```python
df['CouponCode']=df['CouponCode'].fillna('NoCoupon')
```

```python
df['TotalPrice']=df['Quantity']*df['UnitPrice']
```

```python
df.to_csv("Cleaned_Dataset_for_PowerBI.csv", index=False)
```
